```python
import os
p = os.getcwd()
while p != os.path.dirname(p) and not os.path.isdir(os.path.join(p, ".git")):
    p = os.path.dirname(p)
root = p if os.path.isdir(os.path.join(p, ".git")) else os.getcwd()
print("repo root:", root)

gi = os.path.join(root, ".gitignore")
lines = ["data_cache.pkl", "otway_windows.npz", "results_otway/", "*.part", "*.progress.json"]
old = open(gi, encoding="utf-8").read().splitlines() if os.path.exists(gi) else []
new = old + [l for l in lines if l not in old]
open(gi, "w", encoding="utf-8").write("\n".join(new) + "\n")
print(open(gi, encoding="utf-8").read())

if os.path.exists("untitled.txt"): os.remove("untitled.txt")      # jo galti se bani thi
```
